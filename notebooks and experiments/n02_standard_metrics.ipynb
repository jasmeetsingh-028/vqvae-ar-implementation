### FID 

Ti measure quality of generated images,by comparing the distribution of generated images to real images using features froma pretrained Inception network. Lower FID scores indicate better quality and diversity of generated images.

### GFLOPs (network complexity)

GFLOPs (Giga Floating Point Operations) measures the computational complexity of a model. It quantifies the number of floating-point operations that one forward pass of the model requires. Lower GFLOPs indicate a more efficient model, which can be beneficial for deployment on resource-constrained devices.
